```python
## create a simple txt file

import os
os.makedirs("../data/text_files",exist_ok="True")
```

```python
sample_texts={
    "../data/text_files/python_intro.txt":"""Python Programming Introduction
Python is a high-level, general-purpose programming language created by Guido van Rossum and released on February 20, 1991.
Key Features & Design
• Interpreted & Dynamic: Code runs line-by-line without a mandatory preliminary compilation step, and variable types are determined automatically at runtime.
• Readable Syntax: It uses clean indentation and English-like keywords rather than complex curly brackets, making it fast to learn and write.
• Multi-Paradigm: It supports object-oriented, procedural, and functional programming styles.
• Name Origin: The language is named after the British comedy troupe Monty Python's Flying Circus, not the reptile.
Common Uses
• Data Science & AI: Widely used with specialized libraries for machine learning, statistical modeling, and data visualization.
• Web & Software Development: Used to build server-side applications, APIs, and cross-platform desktop software.
• Task Automation & Scripting: Popular among developers and non-programmers alike to automate repetitive workflows, file management, and system tasks.""",
     "../data/text_files/machine_learning.txt":"""Machine Learning Basics
Machine learning is a subfield of artificial intelligence (AI) that allows computers to learn from data and make predictions or decisions without being explicitly programmed for every task.

Core Types of Machine Learning

• Supervised Learning: Trains algorithms using labeled datasets to classify data or predict continuous outcomes (e.g., spam detection, medical diagnosis).
• Unsupervised Learning: Analyzes unlabeled data to find hidden structures, patterns, or groupings like clustering and anomaly detection.
• Semi-Supervised Learning: Uses a small amount of labeled data combined with a large pool of unlabeled data to improve accuracy.
• Reinforcement Learning: Trains an agent to make decisions through a trial-and-error reward and penalty system in a dynamic environment.
• Generative AI: A specialized subset of machine learning that learns data distributions to create new content such as text, images, or audio.

     """    
}

for filepath,content in sample_texts.items():
    with open(filepath,'w',encoding="utf-8") as f:
        f.write(content)

print("Sample text files created!")        
```

```python
### TextLoader

#from langchain.document_loaders import TextLoader

from langchain_community.document_loaders import TextLoader

loader=TextLoader("../data/text_files/python_intro.txt",encoding="utf-8")

document=loader.load()
print(document)
```

```python
### Directory Loader

from langchain_community.document_loaders import DirectoryLoader

##load all the text files from the directory
dir_loader=DirectoryLoader(
    "../data/text_files",
    glob="**/*.txt", ##Pattern to match files
    loader_cls= TextLoader, ##loader class to use
    loader_kwargs={'encoding':'utf-8'},
    show_progress=False
)

document=dir_loader.load()
document
```

```python
###PDF READING

from langchain_community.document_loaders import PyPDFLoader,PyMuPDFLoader
from langchain_community.document_loaders import DirectoryLoader

##load all the text files from the directory
dir_loader=DirectoryLoader(
    "../data/pdfs",
    glob="**/*.pdf", ##Pattern to match files
    loader_cls= PyMuPDFLoader, ##loader class to use
    # loader_kwargs={'encoding':'utf-8'}, //as encoding is not required in pdf
    show_progress=False
)

pdf_documents=dir_loader.load()
pdf_documents
```
